```python
# Funcion de manera iterativa

def conteoRegresivo(contador):
    for i in range(contador,0,-1):
        print(f"*** {i} ***")
    print("¡Despegó!")

conteoRegresivo(5)
```

```python
# Funcion de manera recursiva

def conteoRegresivoRecursivo(contador):
    if contador == 0:
        print("¡Despegó!")
    else:
        print(f"*** {contador} ***")
        conteoRegresivoRecursivo(contador - 1) 

conteoRegresivoRecursivo(5)
```

```python
# Funcion para sumar 1 a n iterativamente

def sumaIter(n):
    result = 0
    for i in range(1,n+1):
        result = result + i
    print(f"La suma de 1 a {n} es: {result}")

sumaIter(5)
```

```python
# Funcion para sumar 1 a n recursivamente

def sumaRecur(n):
    if n == 0:
        return 0
    else:
        return n + sumaRecur(n-1)

print(f"La suma de sus numeros es: {sumaRecur(5)}")
```

```python
# Funcion para invertir una cadena de caracteres iterativamente

def invertirCadenaIter(cadena):
    cadenaInvertida = ""
    for i in range(len(cadena)-1,-1,-1):
        cadenaInvertida = cadenaInvertida + cadena[i]
    print(f"La cadena invertida es: {cadenaInvertida}")

invertirCadenaIter("anitalavalatina")
```

```python
# Funcion para invertir una cadena de caracteres recursivamente

def invertirCadenaRecur(cadena):
    if len(cadena) == 0:
        return ""
    else:
        return cadena[-1] + invertirCadenaRecur(cadena[:-1])

print(f"La cadena invertida es: {invertirCadenaRecur('Hola Mundo')}")
```
