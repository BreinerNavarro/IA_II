```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
data = pd.read_csv("Ice_cream_selling_data.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
x = data["Temperature (°C)"]
y = data["Ice Cream Sales (units)"]

plt.scatter(x, y, label="Datos", color="red")

plt.title("Ventas de Helado según la Temperatura")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Ventas de Helados (1,000 unidades)")
plt.legend()
plt.grid(True)

# Mostrar la gráfica
plt.show()
```
