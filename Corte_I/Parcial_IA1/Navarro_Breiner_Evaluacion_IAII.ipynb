## Modelo de Regresion lineal Multiple

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error
```

```python
data = pd.read_csv('datos_entrenamiento.csv')
data.head()
```

```python
data.describe()
```

```python
data.shape
```

```python
print(data.isnull().sum())
```

```python
plt.scatter(data["id_producto"], data["valor_venta"])
plt.title("Producto vs Precio")
plt.xlabel("Producto")
plt.ylabel("precio")
plt.show()
```

```python
# Matriz de correlacion

corr_matriz = data.corr(numeric_only=True)

sns.heatmap(corr_matriz, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Matriz de correalación")
plt.show()
```

```python
x = data[["costo_produccion", "presupuesto_marketing"]]
y = data["valor_venta"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
y_predict = model.predict(x_test)
r2 = r2_score(y_test, y_predict)
rms = mean_squared_error(y_test, y_predict) ** (1 / 2)
```

```python
print("Coeficientes del modelo : ", model.coef_)
print("Intercepto : ", model.intercept_)
print(f"r2 del modelo :  {r2:.2f}")
print(f"Error cuadratico medio : {rms:.2f}")
```

```python
sns.scatterplot(x=y_test, y=y_predict)
lim = [y_test.min(), y_test.max()]
plt.plot(lim, lim, color="red", linewidth=2)
plt.xlabel("Valor de venta real")
plt.ylabel("Valor de venta predicho")
plt.title("Comparación del valor real con el valor predicho")
plt.grid()
plt.show()
```

## Modelo de Regresion Lineal

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_squared_error
import numpy as np
```

```python
data2 = pd.read_csv("datos_entrenamiento.csv")

x1 = data2["valor_venta"]
y1 = data2["costo_produccion"]

plt.scatter(x1,y1)
plt.title("Valor de venta vs Costo de produccion")
plt.xlabel("Valor de venta")
plt.ylabel("Costo de produccion")
plt.show()
```

```python
X1= np.array(x1).reshape(-1,1)
Y1= np.array(y1)

modelo = LinearRegression()
modelo.fit(X1,Y1)
predic_y = modelo.predict(X1)

r2L = r2_score(Y1,predic_y)
mseL = mean_squared_error(Y1,predic_y)**(1/2)

print("R2:",r2L)
print("MSE:",mseL)


plt.scatter(X1,Y1,c= "red")
plt.plot(X1,predic_y)
plt.grid()
```
