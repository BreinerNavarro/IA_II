```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
digitos = load_digits()
digitos
```

```python
imagen = np.reshape(digitos.data[9], (8,8)) # redimension de la matriz para poderla grafica
```

```python
plt.imshow(imagen, cmap = "gray")
```

```python
x_train, x_test, y_train, y_test = train_test_split(digitos.data, digitos.target, test_size= 0.3, random_state= 42)
```

```python
modelo_RL = LogisticRegression()
modelo_RL.fit(x_train, y_train)
```

```python
predicciones = modelo_RL.predict(x_test)
predicciones
```

```python
from sklearn.metrics import  confusion_matrix
matriz = confusion_matrix(y_test, predicciones)
matriz
```

```python
sns.heatmap(matriz, annot=True, linewidths=0.5, cmap = "coolwarm")
plt.ylabel("Datos realaes")
plt.xlabel("Datos predichos")
```

```python
import pandas as pd

data = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

data.head()
```

```python
data.info()
```

```python
#Transformar las variables objetos a numericas
data.TotalCharges =pd.to_numeric(data.TotalCharges, errors = "coerce")
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace = True) # remplaza esto data = data.dropna()
```

```python
data.isnull().sum()
```

```python
data.drop("customerID", axis = 1, inplace = True)
data.head()
```

```python
data["Churn"].replace(to_replace="Yes", value=1, inplace= True)
data["Churn"].replace(to_replace="No", value=0, inplace= True)
```

```python
data.head()
```

```python
data_processing = data.copy()
```

```python
data_processing = pd.get_dummies(data_processing).astype(int)
data_processing.head()
```

```python
fig = plt.figure(figsize=(15,9))
data_processing.corr()["Churn_Yes"].sort_values(ascending=True).plot(kind = "bar")
plt.show()
```

```python
#Escalizar los datos

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler() #normalización de los datos
data_scaler =  scaler.fit_transform(data_processing)
data_scaler
```

```python
 data_scaler = pd.DataFrame(data_scaler)
 data_scaler
```

```python
data_scaler.columns = data_processing.columns
data_scaler
```

```python
def grafica_cateroricas(columna):
  sns.countplot(data = data, x = columna, hue ="Churn")
  plt.show()
```

```python
columnas_categoricas = data.select_dtypes(include="object").columns
```
