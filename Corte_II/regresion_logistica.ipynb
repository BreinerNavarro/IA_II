```python
import pandas as pd

data = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

data.head()
```

```python
data.info()
```

```python
#Transformar las variables objetos a numericas
data.TotalCharges =pd.to_numeric(data.TotalCharges, errors = "coerce")
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace = True) # remplaza esto data = data.dropna()
```

```python
data.isnull().sum()
```

```python
data.drop("customerID", axis = 1, inplace = True)
data.head()
```

```python
data["Churn"].replace(to_replace="Yes", value=1, inplace= True)
data["Churn"].replace(to_replace="No", value=0, inplace= True)
```

```python
data.head()
```

```python
data_processing = data.copy()
```

```python

data_processing["Churn"] = data_processing["Churn"].map({"Yes": 1, "No": 0})
#data_processing = pd.get_dummies(data_processing, drop_first=True).astype(int)
data_processing = pd.get_dummies(data_processing).astype(int)
data_processing.head()
```

```python
fig = plt.figure(figsize=(15,9))
data_processing.corr()["Churn"].sort_values(ascending=True).plot(kind = "bar")
plt.show()
```

```python
#Escalizar los datos

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler() #normalización de los datos
data_scaler =  scaler.fit_transform(data_processing)
data_scaler
```

```python
 data_scaler = pd.DataFrame(data_scaler)
 data_scaler
```

```python
data_scaler.columns = data_processing.columns
data_scaler
```

```python
def grafica_cateroricas(columna):
  sns.countplot(data = data, x = columna, hue ="Churn")
  plt.show()
```

```python
columnas_categoricas = data.select_dtypes(include="object").columns
```

```python
for columna in columnas_categoricas:
  grafica_cateroricas(columna)
```

```python
# Variables numericas
sns.pairplot(data=data, hue="Churn")
```

```python
x = data_scaler.drop("Churn", axis = 1)
y = data_scaler["Churn"].values
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.3, random_state= 42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
resul = model.fit(x_train, y_train)
```

```python
from sklearn import metrics
predections = model.predict(x_test)
accuracy = metrics.accuracy_score(y_test, predections) # exactitud
print(accuracy)
```

```python
model.predict_proba(x_test) # evaluar la probabilidad de 1 o 0
```

```python
model.coef_ # importancia de cada caracteristica
```

```python
model.feature_names_in_
```

```python
# Grafica las variables con coeficientes mas positivos

pesos = pd.Series(model.coef_[0],index=x.columns.values)

pesos.sort_values(ascending = False)[:10].plot(kind="bar")
```

```python
pesos.sort_values(ascending = False)[-10:].plot(kind="bar")
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix

matriz = confusion_matrix(y_test, predections)

disp = ConfusionMatrixDisplay(confusion_matrix = matriz, display_labels=model.classes_)
disp.plot(cmap=plt.cm.Purples)
```
