```python

import matplotlib.pyplot as plt
            
altura = [173, 171, 189, 181]
peso = [81, 72, 96, 94]


plt.scatter(altura, peso)
plt.grid()
plt.title("Atura vs peso")
plt.xlabel("Altura(cm)")
plt.ylabel("Peso(kg)")
plt.show()

```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
from sklearn.metrics import r2_score, mean_squared_error


# datos 

altura = [173, 171, 189, 181]
peso = [81, 72, 96, 94]


# Convertir los datos en arrays 

X = np.array(altura).reshape(-1,1)
Y = np.array(peso)


# Crear y entrenar el modelo 

modelo = LinearRegression()
modelo.fit(X,Y)


# obtener parametros del modelo 

b = modelo.intercept_
m = modelo.coef_

y_predic = modelo.predict(X)
print(y_predic)
print(Y)


# Calcular metricas 

r2 = r2_score(Y, y_predic)
rmse = mean_squared_error(Y, y_predic)**(1/2)

print("El valor del r2 es : ", r2) # un buen resultado es por encima de 0.8
print("El valor del rmse es : ", rmse) # que tanto se equivoca el modelo en las unidades de y 


```

```python

# graficar modelo con datos realaes 

plt.scatter(altura, peso, color = "blue", s=100, label = "Datos reales")
plt.plot( X, y_predic, color = "red", linewidth = 2, label = f"Modelo regresión lineal (R2 = {r2:.2f})")


plt.title("Modelo de regresión lineal vs datos reales")
plt.xlabel("Altura(cm)")
plt.ylabel("Peso(kg)")
plt.grid()
plt.legend()
plt.show()



```
